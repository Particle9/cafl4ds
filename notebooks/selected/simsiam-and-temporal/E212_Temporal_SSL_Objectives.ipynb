# E212 · Temporal prediction and SimSiam invariance

Nine paired 672-minibatch trajectories compared invariance, a gradient-calibrated 50/50 mix, and temporal prediction. Each frozen DINO target became available when its frame arrived. The temporal evaluation uses separate clips.
