# SimSiam batch-size pilot and MAE-style Lasso

Batch-8 timing branches **108/108**. Consecutive-update branches **9/9**. Status **complete**.
Updated 2026-11-20T17:54:51.832140+01:00.

The batch-size pilot keeps the same 256-frame window, four update times, origin checkpoints and label-free evaluation banks. Batch 8 takes eight evenly spaced frames from each original 32-frame block. The batch-32 and Skip results are reused. Early, middle and late clip positions are balanced over nine states and three source seeds.

The Lasso uses the original MAE formulation, LassoLars with alpha 0.002 per output and training-fold-only scaling. Outputs are SimSiam SSL loss, RankMe and feature variance, plus centered rank, retrieval and LiDAR where recorded. Both feature sets use the observed prefix and historical health. The second adds pre-update SimSiam-specific readouts. Forecasts hold out entire source seeds and all of their clips. Future local/global bank outcomes are targets. The timing-interaction variant lets the coefficient on elapsed time depend on the observed origin. Plain additive Lasso can only learn one ranking of the single-update times per fit.
