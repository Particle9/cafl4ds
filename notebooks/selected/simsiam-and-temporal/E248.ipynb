# E248 · Which label-free proposals yield real object matches?

24 independent BDD100K MOT clips; 0.2 s primary, 0.4 s sensitivity.

Texture, camera-compensated residual motion, and frozen-DINO appearance each place exactly two region proposals per frame. All use the original frozen E223 matcher and cutoff. Only previously observed frames enter matching. MOT annotations enter scoring after all predictions are locked.

Primary outcome is correct tracked-object matches per 100 region proposals. Coverage is conditional on having a scorable tracked source point still visible at the other endpoint. Moving denotes at least 3 pixels of normalized box displacement. Paired intervals resample capture clips. The two primary comparisons use a multiplicity-adjusted 95% family interval.

**Audit gate** · No proposal passed. Passing requires +5 correct matches per 100 proposals over texture and a positive adjusted lower interval.
