# E217 · Temporal prediction beyond persistence

A residual MLP started at exact persistence on frozen DINO embeddings. Epoch and residual shrinkage were chosen on development clips. Training doubled the weight of pairs with large frozen-embedding displacement. The high-change cutoff was estimated per lag on development clips. Positive skill means lower MSE than persistence.

Held-out pooled skill was **+6.68%** overall and **+11.72%** on high-change pairs. Clip-bootstrap intervals were +5.04%, +8.28% overall and +9.42%, +14.25% on high-change pairs.
