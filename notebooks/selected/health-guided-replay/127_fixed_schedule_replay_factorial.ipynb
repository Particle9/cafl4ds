# N127 interpretation: input placement matters, but a net transfer benefit is unqualified

All 432 forks completed, with exactly 32 successful updates each. Frozen source,
program, generator-decision, action-count, causal replay and saved-probability
audits passed. Six independent training histories share one pretrained parent;
the development readout panel and historical project source are reused.

## The primary result is directional, not decisive

Evenly spaced minus first-block mean balanced CE is **−0.001289**, averaging the
three action mixtures and three new input histories within learner first.
Five of six learner means improve. The descriptive learner-cluster 95% interval
is [−0.002672, +0.000068], which includes zero. The three-reader sensitivity
retains the mean direction (−0.001771) but also has an interval spanning zero.

Balanced accuracy rises by only 0.040 percentage points, and mean worst-cell
accuracy by 0.077 points. Both descriptive intervals include zero. This is a
modest CE-oriented signal rather than a broad classification/retention result.

## Most of the placement signal comes from ordinary-only training

| Refresh-plus-replay updates / 32 | Uniform minus front CE | Accuracy change (pp) |
|---|---:|---:|
| 0 | −0.003222 | +0.106 |
| 8 | −0.000507 | −0.002 |
| 32 | −0.000139 | +0.016 |

Ordinary-only training contributes about 83% of the pooled mean placement gain.
Its CE contrast improves in five of six learners and has a descriptive interval
[−0.006006, −0.000431]. The sparse/full combined-action contrasts are much smaller
and their intervals span zero. These are secondary factorial results, not a
replacement primary selected after seeing outcomes.

The interaction is consistent with the combined refresh/replay action reducing
sensitivity to the poor early-input schedule. It does not isolate replay from
momentum refresh, because that action changes both.

## Later inputs remain an alternative to a special spacing effect

On new inputs with ordinary updates, the last-block CE is 0.690940, compared with
0.691898 for evenly spaced and 0.695119 for first-block training. Uniform does not
demonstrate a lead over the last block. In natural continuation, the last block
has the lowest mean CE at every action mixture; random placement does not repeat
its new-input benefit there.

Placement changes which arriving images are trained on, their ordering, and the
available causal replay pool. This is not evidence that elapsed waiting alone
helps. The data support avoiding front-loaded ordinary updates on these new
input histories more strongly than they support a universal evenly spaced or
random schedule.

## Sparse combined actions have a favorable observed cost/CE tradeoff

For evenly spaced new-input training, 8 refresh-plus-replay actions yield CE
0.690771 versus 0.690784 for 32 such actions. Their mean difference is about
0.000013, with a descriptive interval spanning zero. Sparse actions require
72 rather than 192 training-gradient images, and about 2.30 rather than 5.14
operational seconds. The close point estimates do not establish equivalence.

Late ordinary-only training is also competitive: CE 0.690940 in about 1.37 seconds,
with slightly higher mean accuracy and worst-cell accuracy than evenly spaced
sparse training. There is no demonstrated need for refresh-plus-replay on every
update, and no uniquely qualified fixed-policy winner.

## The unchanged-parent reference changes the practical interpretation

A post-hoc saved-array audit verifies that every initial readout for a given
parent is identical across its forks. The unchanged-parent mean CE is 0.689912.
Every one of the twelve new-input placement/mixture conditions has higher mean
CE than that reference. Even the lowest mean new-input CE, evenly spaced sparse
training, is +0.000859 above the initial state. Its small accuracy gain coexists
with this worse CE. The source remains unchanged when no updates execute, so
the initial readout is the appropriate unchanged-model reference; no extra
training run is needed for that comparison.

Thus a gain relative to a poor training schedule is not a demonstrated net CE
benefit from adapting to the new inputs. This is a post-hoc diagnostic comparison,
not the prespecified primary outcome, and its intervals must also be retained.
It does not prove every policy is harmful for every learner or task metric.

Natural continuation differs: most mean CE changes improve relative to the
initial state. Late ordinary-only training has the best observed mean there,
about −0.002636 CE from initial. That net contrast improves in only three of six
learners, and one large improvement contributes more than the entire net mean
gain. Its descriptive interval [−0.010250, +0.003946] includes zero.
Same-history uniform versus front remains
heterogeneous: −0.000604 with 12 readers but +0.001632 with three readers. This
limits a general placement claim across input settings.

## Scope and decision

Three generation histories encountered a qualifying temporary 64-hold event,
but none was stalled at the fixed 6,400-arrival snapshot. All six distinct model
and optimizer states entered the experiment; there was no stall-based selection
or early termination. N127 does not qualify stalled-state recovery.

Keep late ordinary and evenly spaced sparse training as cheap candidate controls.
The immediate qualification question is when an update policy improves upon
leaving the model unchanged, with CE, accuracy and worst-cell outcomes reported
separately. N127 supports no automatic adoption of a spacing rule, full replay,
or greater feedback/critic/horizon complexity.

The unchanged-parent comparison is explicitly post-hoc and uses saved readouts
only. Its reproducible source is `interpret_results.py`, with estimates and
descriptive intervals in `interpretation.json`. No new training was performed.
