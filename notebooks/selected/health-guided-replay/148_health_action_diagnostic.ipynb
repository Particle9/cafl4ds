# N148 — Missing actions, prediction errors, or an immediate objective that does not persist?

N147 completed three streams with exact-budget controls. Its H1 controller
underperformed block-random placement on final historical-bank reconstruction
loss and RankMe in all three streams. N148 diagnoses why before another deployed
controller is trained. Health and SSL reconstruction are the only evaluations:
no labels, linear probes, kNN, fine-tuning, or task heads are loaded.

## Frozen sample and parent reconstruction

Use N147's three adaptive-H1 trajectories, each from the same original pretrained
model and Adam state. Select 24 accepted and 24 skipped arrivals uniformly without
replacement in EACH stream, seeds 148101/148201/148301. Require a complete 256-image
past/current window and 32 subsequent arrivals. Selection uses only the saved
action, measurement flag and arrival index, never loss or health outcomes.
This gives 144 origins. Record eligible population sizes and inclusion probabilities.
Report accepted and skipped strata separately; their 50/50 sample is not the
natural arrival distribution. These are reused development histories and states,
not fresh validation or independent pretrained models.

Replay each original prefix from the original full learner/Adam/RNG parent using
N147's saved actions and per-image training seeds. At every sampled state verify
the current-image loss and rolling RankMe against the original saved pre-action
reading. Replay the factual action afterwards and verify its saved post reading.
Verify the reconstructed final full-state fingerprint against N147. Temporary
branches never change the factual parent trajectory or its predictor.

## Five actual alternatives, two candidate view seeds

From each exact parent, measure hold; current-image steps at 1/16, 1/4 and 1 of
the original nominal LR 1e-4; and a quarter-rate replay step. The replay loss is
0.5 times current-image loss plus 0.5 times the mean training loss of FOUR distinct
past images uniformly sampled from up to the preceding 4096 arrivals. All past
arrivals are eligible, including those originally skipped. Save replay identities
before branching, and verify pixel disjointness from the separate audit bank.
Replay uses one Adam step and five gradient images; ordinary steps use one step
and one image. Preserve Adam moments, clipping at 1.0, weight decay and step
semantics; no momentum reset or displacement rescaling is introduced.

View 0 uses the original N147 candidate seed. View 1 uses a separate predeclared
N148 namespace. All positive actions within a view share the current-image draw;
replay's four-image forward follows the current forward. The same replay identities
are used for both views. Holds are duplicated as deterministic verification, not
independent observations. Total: 144 x 2 x 5 = 1440 candidate branches.

After each candidate, execute exactly the SAME 32 actual subsequent singleton
quarter-rate updates, using their original per-image seeds, across all actions
and both candidate views. Compare each endpoint to hold plus the same continuation.
Call this **after 32 continuation updates** (candidate H1 followed by 32; H33 for
positive candidates), not an H32 rollout. It is one specified continuation policy,
not a claim about the original controller's future actions. Future images are
used only in retrospective branch evaluation, never for an online decision.

## Measurements and diagnostic decisions

Before, immediately after, and after 32 continuation updates, measure actual
reconstruction of the ORIGINAL incoming image and RankMe of the SAME original
256-image window. Fixed identities and assay masks isolate parameter effects
from rolling-window membership changes. Also measure the separate fixed historical
audit bank's loss, RankMe, variance, covariance, norm and drift. At the continuation
endpoint additionally record the actual endpoint image loss and its newly rolling
256-image RankMe; keep this distinct from the original-population persistence assay.
RankMe is a population measurement, never a rank assigned to one image.

Freeze original H1 forecasts and margin-based decisions from the N147 trace.
Compare forecasts with measured quarter/full outcomes at accepted AND skipped
origins. Separate raw-prediction decisions from margin-gated decisions to identify
whether missed measured opportunities arise from predictions or the margins.
Do not extrapolate the old head to 1/16 or encode replay as a scalar learning rate.
No head is trained and no margins are calibrated in this diagnostic.

For each measured menu {0,1/4,1}, {0,1/16,1/4}, {0,1/4,replay}, and the union,
define a retrospective immediate oracle: actual loss gain > .001 AND actual
RankMe gain > 0, then minimize after_loss/before_loss + before_rank/after_rank.
Hold if no positive action qualifies. These are outcome-informed diagnostic
choices, not deployable policies or proven long-horizon upper bounds.

Report measured opportunity frequency, missed opportunities at skipped origins,
false acceptance, immediate choice agreement and constrained-score regret.
For each menu-selected action, report both historical loss and RankMe effects
after the shared continuation, separately. Measure whether immediate feasible
actions preserve historical health after continuation; retain all tradeoffs.
Show agreement across the two candidate views. Do not select a favorable view,
stream, checkpoint or score weight after seeing results.

Report every stream and stratum separately, then equal-weight stream summaries
within each stratum. Pair actions at each origin and average the two views within
origins; origins, views, overlapping continuation windows and shared-parent streams
are dependent. No significance claim based on treating 1440 branches as replicates.
There is no universal winner or automatic next deployment from this experiment.

Interpretation: accurate immediate choices with poor historical persistence
motivate objective/horizon work; existing useful actions missed by forecasts
motivate predictor/calibration work; additional useful, persistent opportunities
from smaller steps or replay motivate a new action menu with new calibration.
These diagnoses can coexist. This study isolates one-action perturbations under
a fixed continuation, not the outcome of repeated oracle deployment.
