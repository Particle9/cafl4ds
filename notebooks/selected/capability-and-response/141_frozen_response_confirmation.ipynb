accuracy or loss since the original pretrained model. MSE is in pp squared.

On the primary same-rate population, geometry reduces frozen-context MSE76.2%

pass the six-comparison adjusted test, with lower MSE-improvement bounds just
above zero. This is genuine prospective evidence of incremental information
in the early response, beyond initial state, action factor and image context.

Neither candidate passes the FULL predeclared primary criterion: adjusted
improvement bounds versus zero and frozen mean still cross zero. Most cases
have small effects and some windows carry much larger harm. These comparisons
have highly variable window-level improvements. Do not describe the result as
either no signal or a fully qualified retention predictor.

## The model is not yet an action-response model

For known states with changed tail rates, geometry is44.0% worse than context
and combined94.5% worse. Combined is also28.8% worse than predicting zero change.
This includes the intended tiny-probe/quarter-tail use case. The new-state
cross-rate stratum is less adverse, but does not rescue the known-state failure.

There is a concrete architectural limitation consistent with this result. The
frozen predictors are additive ridge models. At a fixed probe state, switching
the proposed tail action changes only the rate and rate-squared columns;
state/response features have no interactions with the proposed action. The
quarter-minus-tiny forecast is therefore constant across ALL contexts:

| Predictor | Harm difference pp | Gain difference pp |
|---|---:|---:|
| Context |+0.66463|+0.00087|
| Geometry |+0.36786|+0.03349|
| Combined |+0.15055|+0.06766|

In particular, both signal models assign the same positive quarter-rate gain
advantage everywhere; only constraints/absolute gain thresholds change the choice.
They cannot learn that one state is more rate-sensitive than another. N140
also confounded probe rate with tail rate because they always matched. N141's
crossed physical responses now provide data to separate those effects.

Thus the useful interpretation is a warning about continuing the current type
of training, not a validated forecast of the outcome of changing the action.
The architecture is one reason to repair action conditioning; this experiment
does not isolate every contribution from distribution shift or missing features.

## Better forecast MSE does not yet imply a useful controller

Shadow actions are one decision per reset context, not a full evolving drive.
The geometry policy's mean tail balanced gain is+0.0472pp versus+0.0352pp for
quota-matched random expectation: an advantage of only+0.0120pp. Combined gives
+0.0355pp versus+0.0219pp, an advantage of+0.0136pp. Matching conditions on
realized per-anchor/probe action frequencies; it is an offline expectation,
not a runnable causal matched-budget policy.

Post-hoc paired-window intervals for these gain differences are positive,
but unadjusted and conditional on fixed task readers, states and quotas. They
do not establish a practically material long-drive improvement. Average max-cell
damage improves descriptively by0.0383pp and0.0342pp versus those expectations,
with intervals crossing zero. CE is slightly worse than the quota expectations.

Geometry picks hold/tiny/quarter20/9/99 times; combined18/1/109. The combined
forecaster has slightly lower same-rate MSE, but its shadow policy has smaller
balanced gain and greater average max-cell damage (0.387pp versus0.283pp).
Prediction quality and control value remain distinct.

All128 eight-update probes are paid, including those followed by a hold.
Selected geometry actions would use3,616 updates across4,096 arrivals and
733.375 nominal LR-sum units (17.90% of full nominal). This is far higher dose
than N101/N137's low-dose full-stream policies. Do not transfer this probing
cadence directly to a near-full BDD run or claim efficiency from this study.

## Recommended repair

Keep early geometry response as the primary simple signal and combined as a
benchmark. Use N141's independently crossed probe/tail cases to fit an explicit
action-conditioned response model. Include the observed probe rate/history and
state/response-by-tail-action interactions; make hold exactly zero. Fit task-cell
changes separately so a balanced mean cannot conceal a retention loss. Any
normalization of response by probe dose is a hypothesis to declare and test,
not a proven linear law for Adam.

Freeze that model and compare it with these unchanged predictors on new windows
before another long run. Confirm useful-update preservation, action ranking and
gain over random/static allocation at comparable committed dose, charging probes
and bank measurement. A short budget-matched continuation guard is appropriate
only after this action-transfer problem is addressed. Longer horizons alone do
not repair the present limitation.

All findings remain conditional on correlated highway-stream development data,
reused query/support panels and one pretrained parent. Runtime signal inputs
are unlabeled, but the risk/gain regressions were constructed offline using
labeled task outcomes. No universal label-free retention guarantee is established.
