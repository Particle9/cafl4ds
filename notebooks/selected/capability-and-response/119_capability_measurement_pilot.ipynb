# N119 findings — capability value needs a delayed, qualified target

The bounded study is complete: **480 matched forks**, 24 parent/input situations,
two first-update views, and ten arms. There are 12 distinct original-cohort plant
states and one shared frozen stalled plant state. The latter is diagnostic-only.
All action/replay/continuation pixels came from the original 2,880 offline TRAIN
frames. Independent hashing found zero development support/query pixel overlap
with all 96,000 training-source frames. No evaluation panel, predictive head fit,
live controller change, or trial-weight checkpoint was involved. Saved output
is about 240 MiB, principally reusable embeddings and probabilities.

## There is delayed action headroom, not yet a working predictor

Choosing an action from measured outcomes under one update view, then testing
that action under the other view, gives H32 paired CE changes of
**-0.005395** in the original cohort and
**-0.007940** in the stalled cohort, versus matched
hold continuations. The corresponding fixed ordinary quarter-rate changes are
+0.000176 and
+0.000064. These are copied-action,
label-informed diagnostics, not predictions available to the controller. H32
includes the same 31 real future-image quarter-rate updates in every branch.
It is not a repeated-current-image rollout or evidence from a full training run.

Refresh plus replay at the quarter-reference displacement has a mean H32 CE
change of -0.004769 in the original cohort and
-0.007856 in the stalled cohort. That interaction deserves explicit
modeling; ordinary and inherited-momentum replay behave differently. The
non-ordinary action labels refer to **matched ordinary displacement**, not a
strict native-LR cap: effective LR factors can exceed the quarter reference.

## Mean task CE does not settle the capability/retention contract

For that same combination, original-cohort three-reader versus twelve-reader
benefit signs agree only 50%
of the time. The twelve-draw target was fixed before collection; extra draws reuse
the same support pool, rather than adding independent labeled sources. A small
legacy three-reader result is therefore not interchangeable with the expanded
target. Twelve-reader unanimity is also a stricter event than three-reader
unanimity and must not be compared as if they had the same sampling probability.

The original-cohort combination improves mean CE while its accuracy changes by
-0.066 percentage points and one domain/class
cell's mean CE worsens by +0.002730. In the stalled cohort,
the corresponding accuracy change is +0.266
points. These are development readout changes, not N95/N117 fresh-audit probe
scores. They do not overturn the earlier long-run accuracy result.

## Retention and health still need separate qualification

The unlabeled alignment sensor's macro rank correlation with worst-cell CE damage
is -0.191 in the
original H32 cohort and
0.495 in the stalled
H32 cohort. It has not earned a hard retention budget. An average task gain cannot
automatically waive damage to a protected cell, and a representation distance is
not automatically functional damage.

Hold and decoder-only actions preserve immediate encoder features and task
probabilities exactly. Translation preserves reader predictions within numerical
precision while moving raw rank/spread. These controls support keeping
instantaneous readable capability distinct from decoder reconstruction and raw
geometry. A **future continuation value** may still depend on decoder and Adam
state, because those change subsequent encoder learning.

## Next construction

Keep H2 and fixed incoming context. Learn a small **action-conditioned delayed
capability advantage** from original-parent development forks, with explicit
ordinary/refresh/replay/combined actions and actual update magnitude/optimizer
descriptors. Do not initialize it from the stalled-cohort labels and then call
the same online stream an independent test. Predict group damage separately.
Use numerical emergency checks and a displacement trust region as health controls;
do not turn either the unqualified alignment score or raw geometry maxima into
new hard semantic-retention limits.

First expand original-parent state coverage within the locked TRAIN pool and
freeze the reader/support target and small head specification. Select/assess on
whole histories and retain separate calibration/reporting roles. The present
12 original states do not qualify a rich deployed model. H2 sequence predictions
also need their own qualification; these first-action forks do not identify all
second-action transitions. Reserve new sources for confirmation before another
long controller comparison. The measurement pilot supplies promising action
labels, rather than an already qualified capability value.

Five unit tests and the real-GPU controls passed. Independent reconstruction of
all 480 saved task targets has zero arithmetic error; common-future hold repeats
are exact. Maximum norm-matching error is
0.00287%. Two engineering amendments are
recorded: restoring process-global RNG after copied trials, and accepting
translation-refit probability differences below 1e-7 (observed maximum 3.36e-9).
Neither changed actions, task targets, or reader selection.

## Interpretation boundaries

The oracle in summary.json selects using measured outcomes and is not deployable. Bootstrap intervals resample at most six reused histories; overlapping readers, views, parents and the stalled source are not independent confirmations. Do not compare three-reader/twelve-reader agreement to a ground-truth accuracy measure.

A translation control preserves support-standardized reader probabilities while raw geometry can change. Decoder-only must preserve immediate encoder features and task probabilities exactly. Alignment is fitted on 128 unlabeled bank images and evaluated on the other 128; it has not earned a hard retention budget. Raw/aligned residuals and four-cell effects are saved for qualification.

No predictor, hard constraint, long-term value or live objective is deployed here. H2 with fixed context remains the intended controller, but these first-action forks do not identify arbitrary H2 sequences. Actual future-stream labels must remain distinct from repeated-current-image predictions.
