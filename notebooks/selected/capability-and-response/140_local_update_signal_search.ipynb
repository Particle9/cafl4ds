## What changed relative to N139

The experiment measures local responses rather than cumulative endpoint drift.
There are now useful and harmful continuations:4 contexts gain at least0.25pp
balanced accuracy while no cell loses more than0.5pp;9 contexts lose more than
0.5pp in a task cell. Primary incremental damage8->32 ranges from-0.018 to
3.994pp. Balanced gains range from-0.300 to+0.363pp. The32-step windows and
action range remain a narrow local population; this does not establish useful
adaptation over a full drive.

## The strongest lead is measured early geometry response

Predictors use actual unlabeled responses during eight paid updates to forecast
damage from the next24 updates at the SAME rate, relative to holding at8.
The primary forecast is maximum class/light accuracy loss, with four200-label
refitted support pools. All rates/anchors from a window are held out together.

|---|---:|---:|

Geometry improves on context in all eight window groups; combined does too.
This is not one window producing every improvement. However most severe harm
occurs on quarter-rate continuation of the original parent. Local incremental
responses are more informative here than the absolute recovery levels tested
in N139. The smaller geometry family is the clearest interpretable candidate;
the combined family is the strongest descriptive accuracy benchmark.

Actual training loss/gradient summaries alone are weaker: MSE0.50928 versus
context0.55834. Whitened recovery does not improve on ordinary aligned recovery.
Do not promote either added complexity merely because it is available.

## Why this is still a lead

No family passes the locked Bonferroni-adjusted tests across eight families,
three baselines and three task targets (72 comparisons). With only eight
window groups, group improvements vary substantially and adjusted lower
bounds cross zero. Those gates are retained, not relaxed after seeing results.

Transport across starting models is also weak. Holding out an entire anchor
gives geometry MSE0.91847, barely below zero0.93846. Combined achieves0.68351,
better than zero but much worse than its0.03737 known-anchor/new-window score.
Thus the present evidence supports a state-conditioned response lead, not a
universal forgetting predictor. All starting states share one pretrained parent,
and all video/query populations are reused development data.

Accuracy-gain prediction is much weaker than harm prediction. Combined reduces
gain MSE36.5% versus context, but only12.6% versus zero; adjusted tests still fail.
The clear diagonal pattern in the harm plot should not be interpreted as an
equally good ability to select beneficial updates.

The bank dynamics model also has real forecast skill: combined past responses
reduce future normalized rank/reconstruction/variance MSE by54.1/74.4/65.4%
versus zero change. These are next24-step health forecasts conditioned on
eight measured updates. They do not establish counterfactual action ranking
or a benefit from longer MPC planning horizons.

## Operational limits and next experiment

These are paid response signals, not free current-image admission features.
The first eight updates may already cause harm. Tiny-rate probing followed by
quarter-rate continuation is NOT measured here: probe and tail rates match.
Runtime bank measurements are label-free, but the diagnostic mapping from
signals to task damage is fitted offline using labeled evaluation outcomes.
Costs include bank readouts; no equal-compute or efficiency benefit is claimed.

The most justified next step is to freeze the geometry predictor and combined
benchmark, then confirm them on additional predetermined, nonoverlapping windows.
Include an additional starting state and explicit cross-rate probe/tail branches
before claiming they can choose a different action after a small probe. Keep
all task-cell outcomes, the frozen parent and dose/random controls. Start with
a short continuation guard; no full-drive MPC upgrade is qualified yet.

Even a successful harm forecast would establish only part of the controller:
it must preserve useful learning and outperform matched random/static allocation,
with the initial paid probing cost charged. The present result is a stronger
reason to investigate early response than to keep changing the global J weights.
