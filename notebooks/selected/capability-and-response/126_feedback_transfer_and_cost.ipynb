## Feedback and repair were behaviorally inactive

In every one of the 20 dose-budget parent/input situations, repaired J/H1 selects
refresh-plus-replay on each of the first 32 arrivals, then stops at the update
cap. Its action vectors and final learner/optimizer/RNG states exactly match
fixed replay in all 20 cases. The measured feedback policy therefore supplies
the same training as a much simpler fixed policy.

The original and repaired heads have zero action or validity-mask differences
at all 3,040 recorded feedback decisions. Their actual action vectors and final
states are identical in all 40 paired situations across both resource regimes.
One forced recovery followed by original J/H1 also ends in the same states as
repaired J/H1 in all 40 situations. These paths exercise no behavioral benefit
from the learned first-moment repair or repeated repair-based action selection.
N124's specific blockage finding is not overturned; it is not encountered here.

## Pricing feedback makes the opportunity cost visible

At matched 32-update dose, J/H1 costs approximately 214.3 operational credits and
9.33 online seconds versus 123.0 credits and 5.24 seconds for fixed replay, with
identical final states. The extra selection and health observations add cost
without changing the training sequence on these measured paths.

Under the common 128-credit operational cap, J/H1 can afford 19 refresh-plus-replay
updates, while fixed replay affords 33. New-input CE is 0.687219 versus 0.686973;
the +0.000246 difference is small and heterogeneous. J/H1 is also worse than the
mean transferred program by +0.000980 CE, with a descriptive interval spanning
zero. It has no demonstrated operational-efficiency advantage.

## What the fixed/shuffled controls suggest

Each shuffled donor control preserves its donor's exact 32-update action counts
while changing order and arrival placement. On the new-input average, shuffled
versions lower mean CE relative to their corresponding donors by 0.003310,
0.002711 and 0.004167. This is secondary evidence worth following up: a useful
action mixture need not preserve the donor's detailed temporal program.

The lowest observed CE belongs to a shuffled program, but that ranking is an
exploratory finding, not a qualified winner selected independently of these target
results. Accuracy does not share a clear overall ranking. J/H1 versus the donor
mean changes balanced accuracy by -0.024 percentage points and mean worst-cell
accuracy by -0.193 points under the update cap. No broad semantic-retention claim
follows from these CE differences or the health metrics.

## Scope and next decision

None of the five target learners stalled during continuation to arrival 6,400.
Consequently, N126 answers the approved independent late-state policy comparison;
the intended multi-learner stalled-state recovery question remains unresolved.
It does not falsify N125's one-learner CE finding or N124's specific validity-gate
repair. It shows that their behavioral advantage did not transfer to these late
states, where J/H1 instead behaves as unconditional replay until its budget ends.

For these measured states, there is no reason to pay for J/H1's feedback when it
reproduces a fixed policy. Simple fixed action mixtures and dispersed scheduling
are stronger controls for further testing. A richer critic or longer horizon is
not supported by this result. A stalled-specific experiment first needs a cohort
that actually reproduces the relevant blockage, with its generation rule declared
before measuring recovery outcomes.
