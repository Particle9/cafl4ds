# Interpretation added after collection: continuation-dependent value and invalid forecasts

Analysis date: 2026-10-30. Run: `project/cafl4ds/outputs/continuation-mismatch/20261030-n123`.

**The next priority is to repair and validate the transition model's physical support, then align action value with the continuation actually deployed.** N123 supports continuation-dependent value. It does not establish repeated-current-image rollout as the main failure, nor justify a larger capability critic or longer planning horizon yet.

## Completion and verification

All 36 parent/view bundles completed: **688 forks**, below the 864 upper bound because identical selected plans were measured once. Each fork spans 32 slots. There were 18,670 executed updates and 6,381 additional ordinary-reference trials. Summed fork execution time was about 71.9 minutes, excluding preparation and parent construction.

The independent saved-array analysis checked all 72 locked input hashes, all 688 saved-array hashes, schedule completeness, causal replay indices, image timing, update accounting, and finite measured health. Reconstructing CE and accuracy from saved probabilities at H1, H8, and H32 reproduced the recorded arrays exactly. The largest relative displacement-matching error was 0.02954%, below the 0.5% tolerance. This verifies saved arithmetic and accounting; it is not an independent GPU rerun. Original inputs remain unchanged; no trial weights were committed and no final audit panel was loaded.

## How to read the measurements

The original cohort has six reused training histories, two parent positions per history, and two first-view replicates: **24 parent/views**. The stalled diagnostic uses **one shared frozen N115 learner**, six input histories, and two views: **12 parent/views**, not 12 independently stalled learners.

Actions are hold, ordinary update, first-moment refresh, and refresh plus replay. Positive actions match the ordinary quarter-rate parameter displacement. Two continuations follow the initial intervention: ordinary quarter-rate updates, or a frozen, greedy J/H1 controller. The latter is not a replay of N122's adapting controller.

Below, benefit is matched-hold balanced development CE minus action CE at H32, multiplied by 1,000. Positive is better. Every mode and continuation has its own hold reference; comparisons across modes are differences in intervention effects, not a common-baseline endpoint comparison. The twelve support-reader draws are sensitivity replicates, not independent training histories.

## 1. First-action value depends strongly on the continuation

Original-cohort mean first-action benefit:

| First action | Ordinary future | J/H1 future |
|---|---:|---:|
| Ordinary | -0.101 | -0.555 |
| Refresh | +2.248 | -0.559 |
| Refresh + replay | +4.756 | +0.209 |

The refresh-plus-replay marginal benefit is approximately 96% smaller under J/H1. Across continuations, the H32 best action agrees in only **8/24** parent/views; mean four-action rank correlation is **0.108**. **35/72** non-hold effects change sign. H1 rankings coincide by construction; agreement falls to 15/24 at H8, then 8/24 at H32.

This is partly understandable from future behavior. After an initial hold, the original-cohort J/H1 continuation averages 3.92 refresh and 13.58 refresh-plus-replay actions over the remaining 31 slots. It can supply later interventions that reduce the marginal value of doing one immediately. Future dose, action direction, and feedback all differ; this is not an equal-compute comparison that isolates a single mechanism.

A small marginal effect also does not imply a poor continuation. At the shared stalled state, hold followed by J/H1 ends at CE 0.64835 and accuracy 65.58%, versus CE 0.65376 and accuracy 65.04% for hold followed by ordinary updates.

**Implication:** an action-value target must specify the future policy. A value learned under an ordinary-update tail cannot automatically be interpreted as value under repeated controller decisions.

## 2. Most complete-plan benefit disappears under first-action execution with J/H1

Original cohort, J/H1 continuation:

| Frozen H2 selector | Execute both actions, actual next image | Execute first action only | Fraction retained |
|---|---:|---:|---:|
| J | +0.748 | +0.188 | 25.1% |
| Prior | +0.762 | +0.205 | 26.9% |
| Rich | +0.754 | +0.230 | 30.5% |

For each selector, the descriptive whole-history bootstrap interval for the first-only minus actual-plan CE difference is positive: J [0.000191, 0.000956], prior [0.000081, 0.001074], rich [0.000133, 0.000957]. These are unadjusted descriptive intervals over six reused histories, not confirmatory significance claims.

With ordinary continuation, first-only execution retains **86%, 98%, and 81%**, respectively. Thus failure to commit the second action is not a universal explanation. At the stalled state, the prior's actual-next plan with J/H1 is mildly harmful relative to its matched hold, while its first action alone is beneficial. Forcing commitment to a second action is therefore not the supported repair.

**Implication:** target the first action followed by the intended replanning policy, or use a terminal value conditioned on that policy. Executing only the first action is normal MPC operation; the issue is whether the forecast and terminal objective correctly represent what happens afterward.

## 3. Repeating the current image is not the main bottleneck in this sample

Replacing the repeated current image with the actual next arrival preserved or improved mean selected-plan benefit in the original cohort. Under ordinary continuation, the improvement was 1.669, 0.079, and 1.791 in the displayed units for J, prior, and rich. The J and rich descriptive intervals exclude zero. Under J/H1 the differences are smaller, and all three intervals include zero.

This does not validate repeated-image rollouts generally. It does weaken that specific explanation for the failures observed here and makes continuation alignment a higher priority.

## 4. Post-hoc diagnosis: impossible optimizer forecasts exclude useful actions

The additional diagnosis reconstructs the frozen planner's saved validity decisions. `capability_mpc.states` predicts additive changes to health and optimizer summaries, then rejects sequences with negative optimizer summaries. Context coordinate 15 is the norm of Adam's first moment, defined by `single_image_signal.optimizer_features`.

At every original parent/view, all four first actions have an admitted continuation. At the shared stalled state:

- Refresh is excluded in **12/12** parent/views.
- Refresh plus replay is excluded in **10/12**.
- For these first-action rejections, the offending coordinate is the predicted first-moment norm: it becomes negative. Refresh predictions range approximately from -0.041 to -0.026, which is physically impossible for a norm.

Measured refresh actions execute numerically and often improve development CE. This is evidence of a transition-model/support failure, not evidence that negative norms should be tolerated or that every measured action preserves task retention.

Using measured outcomes only as a nondeployable oracle diagnostic:

| Stalled-state first-action oracle benefit | All four measured actions | Forecast-admitted actions | Benefit removed |
|---|---:|---:|---:|
| Ordinary continuation | +7.425 | +0.392 | 94.7% |
| J/H1 continuation | +1.662 | +0.315 | 81.1% |

The measured best action is excluded in 9/12 and 7/12 parent/views, respectively. Most of this oracle headroom is inaccessible to any score operating behind the current validity mask. The prior and rich selectors choose hold in 10/12 stalled cases and refresh plus replay in two. Treating this primarily as insufficient critic expressiveness would miss the upstream failure.

This diagnosis concerns the frozen N123 parent states. It does not establish that the same mechanism caused every N122 deferral, where models adapted and visited different states.

## Health metrics and the next experiment

Health metrics remain useful observations and predictors of training dynamics, but the results do not support treating their geometry thresholds as established semantic-retention guarantees. For example, original-cohort refresh plus replay under ordinary continuation improves balanced CE by 0.004756 while mean accuracy changes by -0.052 percentage points, and one of four domain/class cells has worse mean CE. Three-reader versus twelve-reader effect signs also disagree in a substantial fraction of cases.

The evidence suggests this sequence:

1. **Repair physical support first.** Compare the frozen additive predictor with positive-support next-state parameterizations for nonnegative summaries, and action-aware treatment of the known optimizer reset. Keep numerical and displacement checks. Simply clipping a negative forecast or removing its validity check would hide the error without validating the forecast. Include late/stalled-state coverage and report validity, action-wise residuals, and false exclusion of measured useful choices.
2. **Measure first-action value under the deployed continuation.** Retain ordinary-tail and frozen J/H1-tail controls, hold references, and matched displacement. Use original-parent states plus genuinely held-out histories for qualification. Keep the shared N115 state a diagnostic stratum. Do not change transition representation and enlarge the critic simultaneously: separate their contributions.
3. **Retain a short-horizon baseline and price the work.** Compare a repaired J/H1 controller with H2 only after useful actions remain available and transition errors are characterized. Count reference trials, candidate probes, replay work, and successful updates; equal displacement does not mean equal computation.
4. **Keep task utility distinct from health.** Health features can inform dynamics, uncertainty, and guardrails. Development CE/accuracy and worst-cell retention should evaluate whether those guardrails serve the task. If labels must remain unavailable online, select and qualify the health-to-utility relationship offline rather than interpreting a geometric floor as a demonstrated task-safety constraint.

There is no demonstrated advantage here that warrants prioritizing the rich capability head. The leave-one-history-out first-action prior favors refresh plus replay under ordinary continuation but becomes harmful on average under J/H1; even a simple action preference is continuation-sensitive. A larger critic or longer horizon should wait for the transition and target definitions to be repaired.
