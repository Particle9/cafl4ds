## Transfer and task benefit remain weaker than the best forecast number

On the new family5009 prefix state, cross-rate harm MSE is0.19151 for geometry,
0.21124 for combined, and0.45555 for zero. Geometry improves its same-data
additive baseline21.7%; combined improves27.4%. No adjusted gate passes there.
Geometry also has lower error than combined on same-rate new-state cases.

Gain forecasting remains less reliable than harm forecasting. On the new state's
cross-rate cases, geometry only slightly improves zero for balanced-gain MSE;
combined is worse. The larger model's best known-state harm score does not
establish a better controller or universal learner-state transfer.

physical LR dose, accuracy and retention are separate tradeoffs.

## Reasonable next step

Use interaction geometry as the primary action-selection candidate, additive
combined as the strong sparse guard comparator, and interaction combined as
a lower-LR/retention comparator if budget permits. Keep random/static allocation
and a frozen-parent reference. A short closed-loop comparison at historical
N101-like dose is more informative now than another unrestricted model sweep.

The local probing cadence itself is not deployable as the next full-drive policy:
selected geometry uses4,904 updates across5,120 arrivals and14.75% of nominal
full LR dose, far above historical low-dose policies. Probe only occasionally,
charge all eight probe updates and bank readouts, and enforce a low LR/update
budget. For example, an eight-tiny-probe/24-step-tail opportunity once per384
arrivals caps quarter-tail LR dose near1.60% of full nominal; this is a proposed
operational hypothesis, not a tested result or selected next protocol.

That pilot must show persistent probe accuracy and per-cell retention while
beating comparable allocation, rather than merely fewer forecast errors.
Point harm<=0.5pp is not a calibrated guarantee:11-14 actual violations remain
for the interaction selectors. A budgeted pilot remains exploratory until its
own prospective efficacy/retention criteria are met; no near-full MPC run is
qualified by this experiment alone.

The video/query populations remain reused development data and all saved
learners share one pretrained parent. Runtime bank inputs are label-free; the
task-response regressions use labeled offline readouts. No live task labels,
source boundaries, future frames or evaluation feedback enter the choices.
