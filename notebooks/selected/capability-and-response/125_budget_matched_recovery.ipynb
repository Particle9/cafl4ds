# N125 interpretation: a J/H1 policy lead on one learner, without a repair effect

The single-learner diagnostic supports keeping the narrow J/H1 policy as a
candidate for further comparison. It supplies evidence that the policy's action
mix and timing can matter beyond simply performing more successful updates.
It does not establish a general controller advantage, a new repair benefit,
task-cell retention, or total-compute efficiency.

## What the behavioral comparison supports

All three arms reach exactly 32 successful updates on each of the three input
histories. Repaired J/H1 lowers mean balanced CE by 0.002001 relative to fixed
refresh-plus-replay and by 0.004725 relative to one-time recovery. Thus its mean
CE lead at this budget is not explained by a higher successful-update count.
The three-reader sensitivity retains both mean effect signs.

Here, one-time recovery means one refresh-plus-replay update followed by ordinary
updates. This study does not test one-time recovery followed by the old J/H1
controller. Its result therefore does not show that repeatedly repairing the
optimizer is necessary.

The J/H1 policy uses a mixture of ordinary and refresh-plus-replay updates and
waits on some arrivals. Fixed replay spends its updates on the first arrivals.
These are policy-level comparisons: action direction, replay dose and timing
change together. There is no matched open-loop schedule control that isolates
the value of feedback itself.

## The repair does not cause this result

A post-hoc direct coefficient decoding compares the original and repaired head
at all 379 recorded controller decisions across both budgets. There are zero
action differences and zero validity-mask differences. All first-arrival actions
are valid under both heads; both choose hold.

Consequently, the original frozen J/H1 controller would select the same actions
along these measured trajectories. N125 does not reproduce N124's diagnosed
optimizer-validity blockage. It provides no additional behavioral evidence for
the first-moment repair itself. With these new inputs, the controller resumes
updates at arrival 3, 3 and 33, respectively, without forcing a recovery action.
The generated stall is context-dependent.

The training-work cap excludes controller gradient probes. At that cap, J/H1
