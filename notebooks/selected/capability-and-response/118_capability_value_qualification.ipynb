## Decision

Neither candidate passes the full development screen.
A qualified deployed capability value has not been established. Do not replace the current MPC objective with these heads or interpret empirical margins as safety guarantees.

First qualify target resolution and reader/support dependence. Low sign agreement weakens single-reader conclusions; it is not an uncertainty estimate for the fixed three-reader mean target. Then collect matched multi-action data at early and stalled modern states: hold, ordinary, matched momentum refresh, replay, and refresh plus replay, with the relevant learning-rate grid. Measure both immediate effects and common real future-stream continuations. Keep every descendant, reader draw, and duplicate parent event on one side of a whole-source split. Lock the measurement/target before new confirmatory data.

For MPC, retain H2 and the fixed-context approximation for immediate dynamics, then add a separately qualified continuation value. Acquisition ranks useful endpoints; retention bounds cell/function damage; health enforces supported emergency limits; work is a separate cost. A predictor trained only on ordinary nominal updates cannot score refreshed/rate-specific transitions reliably.
