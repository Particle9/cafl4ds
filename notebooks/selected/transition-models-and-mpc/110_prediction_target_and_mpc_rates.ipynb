Status: **complete**. Active: **preparation / verification**.

New sequence: direct delta-J H1, then rich fixed-context H2 with physical rates
{0,1/8,1/4} of ORIGINAL nominal LR. Reuse completed verified N109 H1/H2 as the
respective references. Separate panels below show the two matched comparisons.

All use the same 88 rich coordinates, original parent, ordered 91,968 reserved
images, view/action seeds, fixed original J, 5% nonzero exploration, alpha=1,
zero-centered ridge prior, 256-successful-update half-life and unclipped frozen
training-only transforms. All 4,032 offline images stay excluded. Labels enter
readouts only. No completed online outcomes or image weights initialize a run.

The scalar fit uses actual TRAIN-only J(after)-J(before), identical context
scalers and physical quadratic-rate design. Its sole output is delta-J; it has
no metric forecasts. Target normalization uses only valid TRAIN positive actions.

only middle physical rate changes from 1/16 to 1/8. New nominal is original / 4.
H2 propagates metric-derived histories/actions and holds other context fixed.

Each arrival logs forecasts before feedback, applies at most one update and
learns only from its successful chosen positive action. Skip/failed/unchosen or
imagined outcomes never become labels. Forecast RMSE uses positive updates;
the frozen-shadow comparison is paired on each run's actual states/actions.
Different controllers visit different states, so cross-run RMSE is descriptive.
