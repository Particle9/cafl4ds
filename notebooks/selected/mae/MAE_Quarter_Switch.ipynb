## Method

Both branches share frames, masks, bank admissions and ridge 0.03 whitening at its scheduled refreshes. Block MSE gains are positive when switching helps. The final source block retains the archived +1 fallback. Selector scalers and ridge coefficients use the first 50% of block origins. Penalties and thresholds use the next 25%. The final 25% evaluates frozen selectors. Paired 95% intervals resample contiguous 384-arrival windows.

The learned override keeps +16 issuing forecasts and learning from its actual trajectory during quarter commitments.
