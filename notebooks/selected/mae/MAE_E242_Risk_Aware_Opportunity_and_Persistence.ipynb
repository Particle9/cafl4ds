# E242 · Risk-aware action opportunity and persistence

The two operating objectives found opportunities in the original branch receipts. On the
independent health bank, **46.2% of the 13
active loss-first selections** and **0.0% of the
4 active rank-first selections** remained useful at +16. At +48,
the respective rates were 53.8% and 25.0%.
The +48 loss-first utility advantage over fixed replay was -0.970 [-2.191, -0.146] pp across
all draws and +0.044 [-0.226, +0.394] pp among the active selections.
This is an offline audit of a single archived test learner.

## Operating rules

**Loss first.** Minimize the mean of the relative new- and historical-bank loss changes,
with equal weights. An action is useful if it reduces this mean by at least 0.25%,
loses at most 0.50% of new-bank RankMe and increases historical reconstruction loss
by at most 0.50%.

**RankMe first.** Maximize relative new-bank RankMe gain while allowing up to 0.50%
relative deterioration on each reconstruction bank. A gain of at least 0.10% is required
to count as useful.

The rules choose the best measured useful action on one mask draw, with Skip as
the fallback. The opposite draw scores that choice. This tests action opportunity and
realized violations. With two draws per state, the audit has limited information
about per-action violation probabilities.

Fixed replay was locked on development data. Among replay actions, select the action
with highest mean utility under a 20% empirical development violation budget. When
the replay candidates all exceed the budget, use the least-violating replay action
and retain the budget miss in the report. At +16 this is 50% replay for both rules,
with development violation rates of 55.6% for loss first and 56.4% for RankMe first.

## Scope and remaining evidence

The +16 replay action with the highest development utility under loss first is
50% replay. Under RankMe first, ignoring risk favors 25% replay, with a 72.8%
development violation rate. Against that replay action the adaptive rank-first
active selections gave +0.207 [-0.031, +0.444]
pp at the independent-bank +16 endpoint, and all four adaptive active selections
violated a loss budget at that endpoint.

The continuation analysis measures representation health. Semantic transfer after
common continuation is a separate question covered by the earlier E244/E249
[persistence gate](MAE_Plus16_Persistence_Gate.ipynb). This audit fits no predictor.
Its one MAE seed and paired draws support an opportunity diagnostic. A
probabilistic multiobjective model would need independent learner trajectories,
explicit calibration and repeatable advantages over fixed replay.
