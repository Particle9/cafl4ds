## Methods

**Saved B** is N90's online heuristic. Its frozen J predictor sets ordinary admission to 50%, 25% or 10% (fallback 25%); the task head then multiplies it by 1.2 for predicted benefit or 0.8 for predicted harm.

**Constant base** replaces every ordinary J probability bin with **11.8514%**. A task head is trained and logged in shadow, with the same features and development readouts, but never adjusts admission.

**Constant base + online task guidance** uses the same constant base, multiplied by 1.2 for predicted benefit (**14.2217%**) or 0.8 for predicted harm (**9.4811%**), clipped to [5%,95%]. Zero/nonfinite predictions and an uninitialized head leave the base unchanged. The head predicts immediate balanced development soft-error change from 19 pre-update features, learns only after actual accepted updates, initializes on 128 labeled accepted samples (first arrival excluded), and discounts old samples with a 256-update half-life. Neutral and task-guided trajectories match until the first initialized head decision.

The base is fixed before training from saved B's average logged ordinary admission probability. No evaluation score selects it. This matches expected opportunity scale, not exact update counts or computation. Frozen J remains a feature in both task heads, so this isolates removing its probability bins rather than removing all J information.

Both new arms independently restore the original complete warm parent, fresh_00 stream, augmentation seed 5003 and acceptance RNG 835003. They keep singleton reconstruction training, fixed bank, J weights 0.25/0.25, health-driven normal LR/4 and caution LR/16, backup/anchor updates, forced bursts and numerical recovery. Task rollback stays disabled. N92's window controller is not used. Incoming images need no class labels; development labels supply task feedback and alignment features. Final evaluation labels only support reporting.

Primary comparisons are late mean balanced linear accuracy (scheduled 30,000--57,600-arrival checkpoints): task-guided minus constant-base, and task-guided minus B. Actual budgets, subgroup performance, CE, kNN and costs are secondary. This is one reused exploratory history/seed. Curves stop at common observed coverage; dotted lines mark the parent. Matched-update plots are descriptive and do not remove selection/rate differences. No efficacy conclusion until both new arms finish.

Late-checkpoint means so far: B: J bins + task **61.851%**, Constant base (shadow task head) **61.641%**, Constant base + online task guidance **61.540%**.
