# N100: Neither vs full training vs all-feature Online J

Status: **complete**. All three methods cover the same 96,000 images,
starting independently from the original parent with the same stream and seeds.

- **Neither (N99):** Lean's reduced features, without clip boundary/position or
  development-accuracy feedback. Uses 66 original inputs plus log2 proposed LR.
  No screening gradients, optimizer summaries or recent-eight-frame similarity.
  Keeps the initial health anchor and skips development-panel observations.
- **Full training (N95):** admits every image for training.
- **Online J: all features (N96):** original 84 retained inputs plus log2 LR,
  including clip metadata, gradient/update features and development feedback.

The two Online J methods use the same J, fixed admission threshold and online
ridge forgetting half-life. Neither's starting ridge was refitted only on the
original calibration data. Removing feedback can change anchors and LR history.
This is a single-history development comparison on the already inspected audit;
audit outcomes never enter training, anchor decisions or checkpoint selection.
