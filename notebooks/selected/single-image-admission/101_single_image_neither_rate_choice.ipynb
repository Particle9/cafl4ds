# N101: Neither with predicted step size

Status: **complete**. Neither is the main baseline. The new method
uses its same 66 context inputs, original parent, 96,000-image stream and seeds.

Score rates **0, 1/64, 1/16, 1/8, 1/4** of nominal LR using one feature pass.
Choose the lowest predicted immediate delta J; skip predicts exactly zero.
On a fixed 5% of eligible arrivals, explore another permitted nonzero rate.
Only the actual successful update supplies a target. No candidate-model trials.

The ridge uses both rate-times-context and squared-rate-times-context terms
(134 coefficients), so its preferred rate can vary by image and model state.
Initialization assumes linear scaling of Neither's original calibration forecast,
with zero curvature; the original single-rate data do not establish a rate curve.
A coefficient-centered ridge prior and 256-accepted-update forgetting half-life
are fixed. No same-stream outcomes initialize or tune the new predictor.

No clip metadata, development feedback, screening gradients, optimizer summaries
or recent-frame similarity. Health remains an input; it cannot trigger separate
rate modes. The old admission bins and stall bursts are replaced by predicted
rate choice. Numerical rollback and a 64-arrival cooldown remain.

Compare whole methods: rate/admission policy and predictor architecture change.
This is a single-history development comparison on an already inspected audit.
Audit outcomes never enter control or select checkpoints. Optimizing immediate
J does not establish improved long-term task accuracy.
