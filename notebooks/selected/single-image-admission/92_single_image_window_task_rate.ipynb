## Methods

**A (offline)** is the saved N84 heuristic. Its frozen J predictor guides image admission, fixed-bank health adjusts step size, and sparse updates can trigger short bursts. Task-based rollback is disabled.

**B (online)** is saved N90. It adds a discounted ridge predictor of immediate development soft-error change, learned only from accepted outcomes. It multiplies ordinary admission probability by 1.2 for predicted benefit or 0.8 for predicted harm, preserving exploration.

**B + window rates (new)** keeps B's admission logic and matches its first 10,240 arrivals. Then each 256-arrival window chooses caution-mode LR/16 or LR/8. The first 16 windows are balanced and randomized. Later choices use a contextual ridge predictor of the full window's development soft-error change, with 20% uniform random exploration. Lower predicted error change is preferred. Feature scaling is fitted on the initial windows only; learning discounts old evidence with a 32-window half-life. The model learns only after a complete window, from its actual assigned action. Two-window persistence is logged separately. Normal-mode LR/4, forced bursts and numerical recovery retain their original rules. The immediate B predictor uses the proposed rate in its own features.

**Full training** is the saved full-exposure reference and uses more optimizer updates. All methods share the parent/stream; the new run keeps seed 5003, J weights 0.25/0.25, the fixed bank and disabled task rollback. Reconstruction remains the image-training loss. Final evaluation labels never choose actions.

Curves stop at the new committed horizon. Dotted lines mark the parent. This is a single exploratory combined-policy comparison on a reused development history, without a new randomized-rate-only control. Different rates can change later admission counts. Primary descriptive comparison: mean balanced evaluation accuracy at scheduled checkpoints from 30,000 through 57,600 arrivals versus B; endpoint, subgroup retention and compute are secondary.

Late-checkpoint mean so far: B **62.010%**, new **62.066%**.
