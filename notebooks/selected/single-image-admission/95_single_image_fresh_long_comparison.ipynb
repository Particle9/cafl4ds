## New history and untouched audit

This comparison uses **1,000 new BDD train highway videos**, 500 daytime and 500
night, with **96 chronological frames per video** at 16 fps, resized to64x64.
The 96,000 arrivals do not repeat video IDs. The stream follows the inherited
DDDDDNNNNN day/night pattern phase. A shared original warm parent has already
seen the same old576-frame prefix; all three methods independently restore its
complete model, optimizer and RNG. The new augmentation namespace/history is
fresh_long_20261026, seed5003. Independence is established at clip level;
recording-session identities are unavailable. These are temporal video frames,
not96,000 independent labeled keyframes or one continuous camera recording.

Membership was fixed before any new-stream outcomes. Previous output paths,
known-video lists, selections and reference manifests were excluded, along with
the warm partition. Decoder failures can use only fixed ordered reserves before
training, never performance-driven replacements. Initial eligible highway pools
after audit selection: {'daytime': 1171, 'night': 3269}.

The **fresh audit** has400 previously unused labeled keyframes,100 per city/highway
by day/night cell. Audit clips are excluded from the new stream and reserves,
all previous streams and control/query/support IDs. Resized pixels are checked
against existing panels and the bank. Readers fit only the same existing support
sets. No fresh-audit score, label or gradient controls admission, rates, anchors,
hyperparameters or checkpoint selection. The old evaluation panel is secondary.
The existing development bank remains fixed for exact N91 method reproduction;
incoming training frames have no labels supplied to the image-model update.

## Three methods

**Full training:** accepts every image and uses the original nominal AdamW rate,
as in N91's full reference. It has no health/admission controller or task head.
Observers run at reporting checkpoints and do not change training.

**N91 online(B):** reuses the exact existing training function. A frozen predictor
estimates a candidate update's change in J (normalized bank reconstruction plus
0.25 rank and0.25 variance deficit hinges). Ordinary admission is50%,25% or10%,
with25% fallback. A separate online ridge head predicts immediate development
soft-error change from19 pre-update features. Negative forecasts multiply
admission by1.2; positive forecasts by0.8; probabilities are clipped[0.05,0.95].
Initialize on128 accepted labeled updates; ridge10/intercept0.1, fixed prefix
scaling/clipping,256-update forgetting half-life. Predict before admission and
learn after a successful update. The head starts untrained on this history.
Health chooses nominal/4 or nominal/16; forced bursts and numerical recovery
are preserved, and task rollback is disabled.

**Constant-base control:** uses fixed ordinary probability11.85138889%, inherited
from the previous B run's mean probability without inspecting this new history.
It has the same health, rate, burst, rich feature and development-readout recipe
as B, and trains the same online head in shadow, but never lets it change
admission. Frozen J bins have no probability authority. This tests the complete
online admission policy against a simple control at a previously calibrated
opportunity scale; realized update/compute budgets need not match.

All image models learn solely through singleton reconstruction updates. J and
development error guide admission only. The N94 delayed-benefit predictor does
not control this run. Full has a different LR recipe by design; comparisons are
between complete methods, not an isolated admission-only causal effect.

## Where the advantage comes from

The late-mean day/night accuracies are65.82%/59.06% for B and64.88%/54.61%
for constant admission. Most of B's advantage over constant comes from night
images (+4.45 points night, +0.95 day). At the endpoint constant actually has
better day accuracy (66.33% versus65.00%); B has better night accuracy
(58.83% versus52.00%). Full's late means are61.63% day and54.96% night.

The old evaluation panel agrees with B's direction: late means62.84% B,
58.81% full and57.73% constant. On the fresh audit, late kNN means also favour B
(57.38%, versus55.56% full and55.96% constant), although B's final kNN accuracy
57.50% is below the parent's59.67%. The representation improvement is not
uniform across readers. Constant has better late linear CE than B

## J is not equivalent to downstream task quality

At the endpoint constant has lower/better J (0.9754 versus1.0334 for B),
higher RankMe (5.59 versus4.72), higher variance (0.00759 versus0.00405),
and slightly lower reconstruction error (0.83237 versus0.83618), while having
worse balanced linear accuracy (59.17% versus61.92%). The same J ordering
holds for the primary late means. These quantities are useful diagnostics,
but optimizing them more strongly need not improve task accuracy. Making
the J ridge online would not by itself resolve this target mismatch.

## Cost and next experiment

B's recorded feature time is119.49 minutes, observations27.92 minutes,
training14.92 minutes and checkpoint/readout work9.49 minutes, totaling
171.81 minutes. Full records33.24 minutes of training. These counters have
different inclusions (full observers are not in that training total) and are
not directly comparable wall-clock budgets. The constant control deliberately
retains B's expensive feature and shadow computations, so it is not an
optimized cheap deployment baseline. Fewer updates have not made B cheaper
in this implementation.

First isolate the rate effect: compare constant admission and B under a shared
fixed conservative LR, alongside a frozen-J-only arm to isolate the task
multiplier. Preserve B as a reference while testing engineering efficiency
changes. A single online-J predictor remains a reasonable ablation, but this
run provides no basis to expect J-only adaptation to improve task accuracy.
Freeze any next protocol before new outcomes. Do not treat the fresh audit as
untouched after using these results for method selection.
