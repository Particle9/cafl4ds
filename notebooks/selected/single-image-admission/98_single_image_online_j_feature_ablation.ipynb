# N98: cheaper features for Online J

Status: **complete**. Two sequential 96,000-image ablations reuse N96's
parent, stream, seeds, J, controller, LR rules and audit checkpoints.

- **No gradient:** remove the nine gradient/proposed-update/alignment inputs and
  their screening backward pass; omit the unused development reference gradient.
- **Lean:** also remove optimizer summaries and previous-eight-frame similarity.
- **Online J (N96):** preserved baseline with all 84 original retained inputs.

Reduced ridges retain 75 and 68 original coordinates respectively, plus log2 LR.
Each is refitted only on the same original 156 calibration rows, with unchanged
scaling, clipping bounds, threshold and 256-accepted-update forgetting half-life.
Removed inputs are absent from fitting/prediction and are never computed.
Development probes and backup/health-anchor rules remain. No separate task head.

This is a single-history development comparison on the inspected N95/N96 audit.
Audit outcomes never control training. Feature removal changes the predictor,
admissions and LR history; this is a comparison of whole methods.
