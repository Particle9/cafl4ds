# N99: Lean without clip metadata or development feedback

Status: **complete**. Three independent sequential 96,000-image runs
use Lean's configuration, original parent, stream, seeds and audit schedule.

- **No clip metadata:** drop clip boundary and position; 66 ridge inputs + LR.
- **No development feedback:** retain 68 inputs + LR; remove development panel
  observations and accuracy/domain-based backup/health-anchor promotion.
- **Neither:** combine both removals; 66 inputs + LR.

The no-development arms keep the initial health anchor and use only unsupervised
health/rate/burst/numerical control. Their zero development traces indicate
unavailable measurements, not measured zero accuracy. Labeled evaluation remains
separate. Original offline calibration and parent initialization are preserved.
No-development starts with Lean's exact original forecaster; clip removals refit
only the original 156 calibration rows. All retain the fixed threshold and
256-accepted-update online-J half-life. No continuation from Lean's final model.

Single-history development ablations on an already inspected audit. No audit
outcome controls training, anchors or checkpoint selection. Removing feedback
can change health anchors, admissions and LR histories; compare whole methods.
