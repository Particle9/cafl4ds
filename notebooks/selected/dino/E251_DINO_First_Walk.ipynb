# E251 · DINO learnability on the first walk

**Descriptive results from development walk 1.** All four trajectories completed.

Offline DINO was up **+6.34 pp on incoming B** with refitted probes after 6,144 updates. At that checkpoint, it was **-3.21 pp on historical A**, **-18.66 pp on B with the frozen probes**, and **+1.90 pp on moving-instance retrieval**. The independent confirmation walks will determine whether either pattern generalizes.

## Protocol and scope

This walk contains 6,144 incoming frames from 12 captures. Replay samples an already observed 64-image bank from 2 historical captures. The offline arm shuffles the same incoming pool and takes a second pass at 6,144 updates. All four arms start from pretrained DINO and use the same cosine EMA schedule.

The semantic development panel uses 48 captures per class and domain across 453 distinct captures. Balanced accuracy gives each class equal weight within a task. Retrieval and semantic labels are confined to evaluation.

Observed trajectory time including readouts on edgelab was Full incoming 11.3 min, Final block + norm 5.2 min, 25% historical replay 11.3 min, Offline shuffle + revisit 22.3 min. The acquisition gate uses independent confirmation walks and query captures.
