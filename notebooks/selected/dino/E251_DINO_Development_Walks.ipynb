## Protocol and provenance

Each walk used 6,144 incoming B frames in chronological online streams with two images per update. The replay policy replaces one incoming image on every second update with an image from an already observed 64-image A reservoir. Offline DINO shuffles the incoming frames and sees the same pool twice at 6,144 updates. All arms start from the same pretrained checkpoint and share the 6,144-update cosine EMA schedule.

The semantic panel contains 453 distinct development captures, with 48 captures per class and domain. Semantic and tracking labels are confined to evaluation.
