## Frozen-probe audit

## Design notes

Online arms consumed B-like frames once in chronological order with a two-image batch. The 25% replay arm substituted past-only A-like images in half of the batches from a 64-image historical reservoir. All arms shared the original DINO start, crops and learning rate. Their common 6,144-update cosine EMA clock kept matched checkpoints comparable. Offline reordered all of the same B frames and revisited them once; 3,072 updates gives the matched comparison and 6,144 tests the more generous ceiling. Readouts use EMA teacher embeddings in each arm. The frozen and refitted probes use the same fit corpus. The tracking benchmark retrieves the same moving instance from same-category candidates in a later view.

## Supervised diagnostic

Last-block label-supervised adaptation on the probe fit corpus produced +0.87 pp on refitted B and +0.00 pp on moving-instance retrieval. It had no role in SSL training or policy selection.
