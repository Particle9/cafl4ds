## Paired design

Training seeds 4, 5 and 6 replay the exact E204 streams 1, 2 and 3, respectively. Each seed pairs all three policies on the same data order and augmentation seed for each update. Every arm starts from the same released weights, heads and primed center. Reference uses the native cosine EMA. Always Fast scales the teacher update rate by 4. Shift-pulsed Fast does so for the first 16 updates after each predeclared boundary. Known boundaries define a scheduling benchmark.

Both networks are measured at the boundaries, +8 and +16 after each shift, and at the endpoint. External labels are used for the same held-out probes as E204. Time averages use trapezoidal integration between measurements. Historical-retention readouts compare the same fixed regime banks to their scores at the end of the latest exposure.
