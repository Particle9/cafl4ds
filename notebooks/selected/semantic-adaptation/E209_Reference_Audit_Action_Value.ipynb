## Paired action-effect diagnostic

Within each source checkpoint, action-effect differences equal endpoint differences because the common origin cancels. For each source seed and held draw, the health ranking uses only the other two draws for that checkpoint. The five-task external mean is scored on the held draw. Only other seeds select the best fixed action. Signs are predeclared: higher RankMe, LiDAR and retrieval margin, lower relational KL. Seed averages receive equal weight. The screening thresholds are 55% concordance, above 50% on at least two seeds, and positive mean value over best fixed. The archived DINO features permit corrected E210-style KL against an independent pretrained reference and against the shared checkpoint, on the same 32 regime-A images with frozen-DINO mutual-neighbor confidence weights. The five external probe banks use different images. The DINO archive supports shared-checkpoint references. It lacks the exact last regime-exit states for these branches. The SimSiam archive supplies scalar health and probe outcomes at +32. It lacks branch endpoint embeddings for relational rescoring.

## Practical effect

The two relational scores gained +0.008 and +0.023 percentage points over the held-out-seed fixed comparator for the DINO student. 0 matched DINO student pairs had a semantic gap of at least 0.5 points. The gains are exploratory at this scale. The SimSiam comparison ended after the archive analysis.
