The study asks whether it matters *when* a learner with a fixed update budget trains. Each walk starts from a common pre-B MAE checkpoint and observes four chronological blocks, B1, B2, A1 and A2. The 192-update arms contain six choices of two full blocks and the evenly spaced schedule. A matched 50% replay schedule challenges the 25% replay policies, and 384 updates at 25% replay provide a full-budget reference.

Every policy admits all observed frames to a 64-image uniform reservoir, including skipped blocks. Training uses the common MAE learning rate and mask seeds paired by potential update position. A block can be buffered before a decision. Later blocks and evaluation banks are unavailable to the selector.

Stage A reuses eight E235 capture-disjoint walks with fresh paired draws. The primary score equally weights A and B and averages four external linear probes within each side (scene, weather, traffic light, pedestrian). Labels serve only retrospective evaluation. The oracle chooses among complete two-block policies using three draws and scores that choice on the fourth. The strongest fixed 192-update policy includes the 50% replay comparator. The interval resamples complete walks.
