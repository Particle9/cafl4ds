# E211 · Plasticity challenge and recovery

Three paired Late4 SimSiam seeds supplied 12 states. Within each seed the challenge used the same novel support clips across checkpoint ages and a separate query bank. Positive values are reductions in fixed-query SimSiam loss. The 72 recovery branches used two draws per state and measured outcomes after 32 and 128 updates.

The early checkpoint improved query SSL loss by **0.0308 more** than the late checkpoint on average at 32 support updates. The loss-of-plasticity screen passed.

| Checkpoint | Query SSL gain at +32 | Five-probe gain at +32 |
| --- | --- | --- |
| 32 | +0.02378 | +0.22 pp |
| 104 | +0.00059 | -0.27 pp |
| 320 | -0.00358 | -0.06 pp |
| 520 | -0.00700 | +0.05 pp |

Mean independent-draw gain over the best fixed action was **+0.11 pp**. The paired seed effects were -0.09 pp, +0.16 pp, +0.28 pp. The gate required +0.50 pp on average and gains on at least two seeds.
