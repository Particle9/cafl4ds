## Which continuation helps?

Each branch starts from the same checkpoint within a state. Standard SimSiam continues with 32 same-image pairs, heads-only freezes the entire backbone and keeps the heads training, and switching uses 32 same-object cross-frame pairs. Every policy is measured after 32 updates. From +128, continue and switch are also measured through +256 on the same continuation. The primary score is the mean of the two balanced accuracies.

Each oracle choice is made using one continuation draw and scored on the other. The best static action is selected from other seeds. The +128 rule switches only at that checkpoint in the local comparisons. The long-horizon comparison measures the actual fixed switch through +256. This is a retrospective diagnostic.
