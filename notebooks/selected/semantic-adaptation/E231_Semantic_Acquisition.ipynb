# E231 · Where does semantic acquisition fail?

**Complete** · E231 finished · Phase C gate stayed closed  
Updated 2026-11-23T11:47:14+01:00 Stockholm

Eight fresh A → B walks are built from capture-disjoint raw BDD videos. Training A and B clips are allocated using frozen MAE features. Each walk has separate raw A and B evaluation clips. The semantic probes score new annotated BDD validation keyframes, with fixed probe fits from earlier studies. The B-matched panel uses the half of the new keyframes most similar to the raw B holdout relative to A. The broad panel uses all the same keyframes. This tests sensitivity to the evaluation domain. The keyframe panels can overlap between walks, so walk intervals condition on this evaluation corpus.

The 96-step replay replication uses one incoming and one historical image from a 64-image earlier-only bank each step. Repetition uses exactly the same incoming images and puts an independently masked copy into the replay slot. The primary score is the mean balanced accuracy of five tasks, with scene kNN shown separately. Labels never enter stream construction, training or selection. A missing task class makes the five-task result unavailable for that walk.
