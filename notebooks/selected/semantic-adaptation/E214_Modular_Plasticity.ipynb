# E214 · Modular plasticity and adapter reuse

Nine paired trajectories compared a shared adapter, fixed routing across four smaller adapters, and nearest-prototype reuse. The router uses frozen DINO embeddings.

Scores are time-averaged balanced accuracy, averaged across three seeds. Times are per trajectory.

The full-trajectory stage screen **did not pass**. Stage-specific policy choices selected on the other seeds gained +0.22 pp over their best fixed policy. Paired seed effects were +0.24 pp, +0.01 pp, +0.40 pp.

Each policy had **196,608** adapter parameters. The reuse policy opened 1, 1, 1 module(s) across the three seeds. The expansion result applies to this router configuration.
