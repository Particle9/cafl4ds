# E224 · Does health feedback add value beyond a good schedule?

6 of six walk/mask groups completed. Each walk contains 1,536 arrivals from sixteen chronological capture-disjoint BDD clips. Within each group, all policies share a calibration parent, images and mask draws. draws use independent mask seeds. Score comparisons average the two mask draws within each walk before comparing the three walks. These clips are absent from predictor fitting, probe banks and the previous holdout.

Every policy has the same count of Skip, 0.25× and 1× actions (384, 363, 21 per 768-arrival block before any whole-walk shuffle). Fixed 50% updates on alternating arrivals. Each adaptive policy ranks only actions whose prespecified block quota remains. The whole-walk shuffle preserves total counts. The local shuffles preserve their source policy’s exact counts within each 96-arrival clip.

The frozen Ridge students fit eight paired +16 outcomes from 1,430 earlier states. Their common base sees elapsed time and recent chosen actions. Image statistics use the current and previous images. Health features are refreshed label-free every 64 arrivals. Both live MLPs use the original rich feature and feedback recipe, with initial weights fitted on the same earlier states for their respective horizon. The +16 reference continuation is fifteen quarter-rate updates. The +1 action value is read immediately. External labels enter only the probes.
